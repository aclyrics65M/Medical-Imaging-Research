```python
from pathlib import Path
from collections import Counter

DATA_ROOT = Path("data")

image_extensions = {
    ".png", ".jpg", ".jpeg",
    ".tif", ".tiff", ".bmp"
}

images = [
    p for p in DATA_ROOT.rglob("*")
    if p.is_file() and p.suffix.lower() in image_extensions
]

print("Dataset directory:", DATA_ROOT.resolve())
print("Total image files:", len(images))

print("\nTop-level contents:")
for path in sorted(DATA_ROOT.iterdir()):
    print("[DIR]" if path.is_dir() else "[FILE]", path.name)

print("\nFirst 15 image paths:")
for path in images[:15]:
    print(path)

print("\nImage counts by parent folder:")
counts = Counter(p.parent.name for p in images)

for folder, count in counts.most_common(20):
    print(f"{folder}: {count}")

print("\nMetadata files:")
for p in DATA_ROOT.rglob("*"):
    if p.is_file() and p.suffix.lower() in {".csv", ".json", ".xlsx"}:
        print(p)
```

```python
from pathlib import Path
import json
import pandas as pd

PROJECT_DIR = Path.cwd()

DATA_ROOT = (
    PROJECT_DIR / ".." / ".." / ".."
    / "Datasets" / "Brain Tumor 12K Mri Images"
).resolve()

IMAGE_ROOT = DATA_ROOT / "Images_" / "Images_"
MANIFEST = PROJECT_DIR / "data" / "manifest.csv"

with open(DATA_ROOT / "DATA.json", "r", encoding="utf-8") as f:
    metadata = json.load(f)

print("Dataset found:", DATA_ROOT.exists())
print("Image directory found:", IMAGE_ROOT.exists())
print("Metadata records:", len(metadata))
```

```python
records = []

for relative_path, info in metadata.items():

    # Original MRI image, not mask or bounding box
    image_file = IMAGE_ROOT / Path(relative_path)

    if not image_file.is_file():
        continue

    lesion = info.get("has_lesion")

    # Only use records with explicit binary labels
    if lesion not in (0, 1, False, True):
        continue

    label = "tumor" if lesion == 1 else "no_tumor"

    records.append({
        "image_path": str(image_file.resolve()),
        "label": label,
        "tumor_type": info.get("tumor_type"),
        "sequence": info.get("sequence"),
        "patient_id": info.get("patient_id")
    })

manifest_df = pd.DataFrame(records)

print("Total valid MRI images:", len(manifest_df))
print("\nClassification labels:")
print(manifest_df["label"].value_counts())

print("\nMissing patient IDs:")
print(manifest_df["patient_id"].isna().sum())

display(manifest_df.head())
```

```python
MANIFEST.parent.mkdir(parents=True, exist_ok=True)

manifest_df.to_csv(MANIFEST, index=False)

print("Manifest saved successfully!")
print("Location:", MANIFEST)
print("Total rows:", len(manifest_df))
```

```python
check_df = pd.read_csv(MANIFEST)

print("Manifest rows:", len(check_df))
print("Columns:", check_df.columns.tolist())
print("\nLabels:")
print(check_df["label"].value_counts())

print("\nPatient identifiers available:")
print(check_df["patient_id"].notna().sum())

display(check_df.head())
```
